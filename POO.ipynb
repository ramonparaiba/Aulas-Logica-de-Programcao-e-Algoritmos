```python
class Aluno:
  def __init__(self, nome, ra, curso):
    self.nome = nome
    self.ra = ra
    self.curso = curso
  def apresenta(self):
    print(f'Sou {self.nome}, {self.ra}, curso {self.curso}')
```

```python
aluno1 = Aluno('Joaquim', '1452142', 'ADS')

print(aluno1.nome)
print(aluno1.ra)
aluno1.apresenta()
```

```python
aluno2 = Aluno('Joaquim', '1452142', 'ADS')

print(aluno1)
print(aluno2)

print(aluno1 == aluno2)
```

```python
#Crie uma classe animal com os atributos:
#nome, especie
#método emitir_som()
```

```python
class Animal:
  def __init__(self, nome, especie, som):
    self.nome = nome
    self.especie = especie
    self.som = som
  def emitir_som(self):
    print(self.som)
```

```python
#Crie uma classe Carro com os atributos
# marca, modelo e ano. e um método descricao()
#que imprime uma frase apresentando o carro.
#insira outro método chamado freiar()
#quando for chamado imprime "O carro freiou"
```

```python
class Carro:
  def __init__(self, marca, modelo, ano):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano

  def descricao(self):
    print(f'Marca:{self.marca}, modelo:{self.modelo}, Ano: {self.ano}')

  def freiar(self):
    print('O carro está freiando!')
```

```python
carro1 = Carro('Fiat', 'Marea', '2009')

carro1.descricao()
carro1.freiar()
```

```python
class Animal:
  def __init__(self, nome):
    self.nome = nome
  def emitirSom(self):
    print("Som genérico de animal")
```

```python
class Cachorro(Animal):
  def emitirSom(self):
    print("Au Au!")

class Gato(Animal):
  def emitirSom(self):
    print("Miau Miau")
```

```python
dog = Cachorro("Charel")
cat = Gato("Carniça")

dog.emitirSom()
cat.emitirSom()
```

```python
#Criar uma classe Funcionario
#Criar subclasses Gerente e Vendedor
#A classe Funcionário tem atributos
# nome  e salário
# método calculaSalário()
#A classe Gerente tem o nome do setor
```

```python
class Funcionario:
  def __init__(self, nome, salario):
    self.nome = nome
    self.salario = salario

  def calculaSalario(self):
    print(self.salario)
```

```python
class Gerente(Funcionario):
  def __init__(self, setor):
    self.setor = setor
  def calculaSalario(self):
    print(self.salario * 1.5)
```

```python
gerente = Gerente("Logística")
gerente.salario = 5000
gerente.calculaSalario()
```
