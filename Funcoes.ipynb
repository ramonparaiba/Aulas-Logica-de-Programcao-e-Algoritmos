```python
#Fórmula IMC = peso / altura**2
def IMC(peso, altura):
  return peso / altura**2


def classificaIMC(peso, altura):
  valorIMC = IMC(peso, altura)
  if (valorIMC > 30):
    return 'Obeso'
  elif (valorIMC > 25):
    return 'Sobrepeso'
  elif (valorIMC > 18.5):
    return 'Eutotrófico'
  else:
    return 'Magro'
```

```python
peso = float(input('Digite o peso: '))
altura = float(input('Digite a altura: '))
print (classificaIMC(peso, altura))
```
